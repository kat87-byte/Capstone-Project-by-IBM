#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
critical_columns = ['Employment', 'JobSat', 'RemoteWork']

print("Missing values before cleaning:")
for col in critical_columns:
    missing_count = df[col].isnull().sum()
    missing_pct = (missing_count / len(df)) * 100
    print(f"Column '{col}': {missing_count} missing values ({missing_pct:.2f}%)")

for col in ['JobSat', 'RemoteWork']:
    df[col] = df[col].fillna('Unknown')

print("\nMissing values after cleaning:")
print(df[critical_columns].isnull().sum())
```

```python
for col in ['JobSat', 'RemoteWork']:
    df[col] = df[col].astype(object)
    df[col] = df[col].fillna('Unknown')


print("\nMissing values after cleaning:")
print(df[critical_columns].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')

df['JobSat_numeric'] = pd.to_numeric(df['JobSat'], errors='coerce')

bins = [-1, 5, 10, 20, 100]
labels = ['0-5 years', '5-10 years', '10-20 years', '>20 years']
df['ExperienceRange'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)

summary_df = df.groupby('ExperienceRange')['JobSat_numeric'].median().reset_index()
print("Median Job Satisfaction by Experience Range:")
print(summary_df)

plt.figure(figsize=(8, 5))
sns.barplot(data=summary_df, x='ExperienceRange', y='JobSat_numeric', hue='ExperienceRange', palette='viridis', legend=False)
plt.title('Median Job Satisfaction by Professional Coding Experience')
plt.xlabel('Experience Range (YearsCodePro)')
plt.ylabel('Median Job Satisfaction Score')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
plot_data = df[df['JobSat'] != 'Unknown']

order = sorted(plot_data['JobSat'].unique())

plt.figure(figsize=(10, 6))
sns.countplot(
    data=plot_data, 
    x='JobSat', 
    order=order, 
    hue='JobSat', 
    palette='magma', 
    legend=False
)

plt.title('Distribution of Job Satisfaction Levels')
plt.xlabel('Job Satisfaction Score')
plt.ylabel('Number of Respondents')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
clean_df = df[(df['RemoteWork'] != 'Unknown') & (df['Employment'] != 'Unknown')]

plt.figure(figsize=(8, 5))
sns.countplot(
    data=clean_df, 
    x='RemoteWork', 
    hue='RemoteWork', 
    palette='Set2', 
    legend=False
)
plt.title('Distribution of Remote Work Preferences')
plt.xlabel('Remote Work Type')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

cross_tab = pd.crosstab(clean_df['Employment'], clean_df['RemoteWork'], margins=True, margins_name="Total")
print("Cross-tabulation of Remote Work Preferences by Employment Type:")
print(cross_tab)
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
lang_df = df.dropna(subset=['LanguageHaveWorkedWith', 'Country']).copy()

lang_df['Language'] = lang_df['LanguageHaveWorkedWith'].str.split(';')
lang_exploded = lang_df.explode('Language')

top_countries = lang_exploded['Country'].value_counts().nlargest(5).index
top_languages = lang_exploded['Language'].value_counts().nlargest(10).index

filtered_lang = lang_exploded[
    lang_exploded['Country'].isin(top_countries) & 
    lang_exploded['Language'].isin(top_languages)
]

ct = pd.crosstab(filtered_lang['Country'], filtered_lang['Language'], normalize='index') * 100

plt.figure(figsize=(12, 6))
sns.heatmap(data=ct, annot=True, fmt=".1f", cmap="YlGnBu", cbar_kws={'label': '% of Developers'})
plt.title('Top Programming Languages Popularity by Region (%)')
plt.xlabel('Programming Language')
plt.ylabel('Country')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
df['YearsCodePro_num'] = pd.to_numeric(df['YearsCodePro'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')
df['JobSatPoints_1_num'] = pd.to_numeric(df['JobSatPoints_1'], errors='coerce')

corr_data = df.dropna(subset=['YearsCodePro_num', 'JobSatPoints_1_num'])

correlation = corr_data['YearsCodePro_num'].corr(corr_data['JobSatPoints_1_num'])
print(f"Pearson correlation coefficient between Experience and Satisfaction: {correlation:.4f}")

plt.figure(figsize=(10, 6))
sns.regplot(
    data=corr_data, 
    x='YearsCodePro_num', 
    y='JobSatPoints_1_num', 
    scatter_kws={'alpha':0.3, 'color': '#1f77b4'}, 
    line_kws={'color': 'red', 'linewidth': 2}
)

plt.title(f'Correlation Between Experience and Job Satisfaction (r = {correlation:.2f})')
plt.xlabel('Years of Professional Experience (YearsCodePro)')
plt.ylabel('Job Satisfaction Score (JobSatPoints_1)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
clean_df = df.dropna(subset=['EdLevel', 'Employment']).copy()
clean_df = clean_df[(clean_df['EdLevel'] != 'Unknown') & (clean_df['Employment'] != 'Unknown')]


clean_df['Employment_clean'] = clean_df['Employment'].str.split(';')
df_exploded = clean_df.explode('Employment_clean')

df_exploded['Employment_clean'] = df_exploded['Employment_clean'].str.strip()

top_employment_types = df_exploded['Employment_clean'].value_counts().nlargest(5).index
df_filtered = df_exploded[df_exploded['Employment_clean'].isin(top_employment_types)]

edu_cross_tab = pd.crosstab(df_filtered['EdLevel'], df_filtered['Employment_clean'], normalize='index') * 100

print("Cleaned Cross-tabulation of Employment Type by Educational Background (%):")
print(edu_cross_tab.round(2))

plt.figure(figsize=(12, 7))
edu_cross_tab.plot(kind='barh', stacked=True, colormap='tab10', edgecolor='black', ax=plt.gca())

plt.title('Employment Type Distribution by Educational Background (Cleaned)')
plt.xlabel('Percentage (%)')
plt.ylabel('Educational Background (EdLevel)')

plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.subplots_adjust(left=0.35, right=0.7)

plt.show()
```
