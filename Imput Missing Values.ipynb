### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path ="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/UDKAZw-kz18Yj8P6icf_qw/survey-data-duplicates.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
print("Number of duplicate rows:", df.duplicated().sum())

duplicate_rows = df[df.duplicated()]
print("\nDuplicate rows:")
print(duplicate_rows)
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
df.drop_duplicates(inplace=True)

print("Remaining rows after removing duplicates:", df.shape[0])
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
missing_remote_work = df['RemoteWork'].isnull().sum()
print("Missing values in 'RemoteWork' column:", missing_remote_work)
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
remote_work_counts = df['RemoteWork'].value_counts()
print("Value counts for RemoteWork:")
print(remote_work_counts)
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
majority_value = df['RemoteWork'].mode()[0]
print("The most frequent (majority) value is:", majority_value)
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
df['RemoteWork'].fillna(majority_value, inplace=True)

print("Missing values in RemoteWork after imputation:", df['RemoteWork'].isnull().sum())
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
comp_columns = [col for col in df.columns if 'comp' in col.lower() or 'salary' in col.lower()]
print("Compensation-related columns found:", comp_columns)


if comp_columns:
    print("\nDistribution summary:")
    print(df[comp_columns].describe())
else:
    print("\nNo compensation-related columns found automatically. Please check df.columns manually.")
```
